## Goal
Load the shared raw MotionSense-AI dataset and inspect its basic structure.

## MotionSense-AI connection
We now turn raw smartphone sensor data into a cleaner and more useful dataset for later visualization and machine learning.

```python
from pathlib import Path

import numpy as np
import pandas as pd

# All notebooks use the same shared data folder.
DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
PROCESSED_DIR = DATA_DIR / "processed"

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Why this matters
Before cleaning or transforming data, we first need to understand what is inside the file.

We will check:
- number of rows and columns
- first few records
- column names
- data types

```python
# Load the raw dataset from the shared data folder.
df = pd.read_csv(RAW_PATH)

# shape returns (number_of_rows, number_of_columns).
print("Dataset shape:", df.shape)

# head() displays the first five rows.
# This gives us a quick preview without printing the full dataset.
df.head()
```

```python
# Display all column names.
# This helps us understand what information is available.
print(df.columns.tolist())
```

```python
# info() shows column names, non-null counts, and data types.
# It is one of the first functions to run on any new dataset.
df.info()
```
