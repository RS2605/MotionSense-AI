# 14 — Filter Activity Rows

## Goal
To use conditions to keep only the rows needed for a specific question.


```python
from pathlib import Path

import numpy as np
import pandas as pd

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
PROCESSED_DIR = DATA_DIR / "processed"

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Filtering
Filtering selects rows that match a condition.

Common examples:
- keep only Walking rows
- keep one device
- keep selected activities

```python
df = pd.read_csv(RAW_PATH)

# Keep only Walking rows.
walking_df = df[df["activity"] == "Walking"]

print("Original shape:", df.shape)
print("Walking shape:", walking_df.shape)
walking_df.head()
```

```python
# Keep rows where the device is phone_A.
phone_a_df = df[df["device"] == "phone_A"]

print("phone_A rows:", phone_a_df.shape[0])
phone_a_df.head()
```

```python
# isin() keeps rows where the value is in a list.
selected_activities = ["Walking", "Standing"]
selected_df = df[df["activity"].isin(selected_activities)]

print(selected_df["activity"].value_counts())
selected_df.head()
```

## Summary
Filtering turns a large dataset into a focused dataset for one research question.
