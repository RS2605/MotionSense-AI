# 19 — Create Features and Encode Activity

## Goal
To create useful sensor features and convert activity labels into numbers.

```python
from pathlib import Path

import numpy as np
import pandas as pd

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
PROCESSED_DIR = DATA_DIR / "processed"

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Feature engineering
Feature engineering means creating new useful columns from existing data.

Here, we create overall movement features from X, Y, and Z sensor readings.

```python
df = pd.read_csv(RAW_PATH)
project_df = df.copy()

sensor_columns = ["acc_x", "acc_y", "acc_z", "gyro_x", "gyro_y", "gyro_z"]

# Fill missing sensor values before creating features.
for column in sensor_columns:
    project_df[column] = project_df[column].fillna(project_df[column].mean())

project_df[sensor_columns].isnull().sum()
```

```python
# Create accelerometer magnitude.
# This combines acc_x, acc_y, and acc_z into one overall movement value.
project_df["acc_magnitude"] = np.sqrt(
    project_df["acc_x"]**2 +
    project_df["acc_y"]**2 +
    project_df["acc_z"]**2
)

# Create gyroscope magnitude.
# This combines gyro_x, gyro_y, and gyro_z into one overall rotation value.
project_df["gyro_magnitude"] = np.sqrt(
    project_df["gyro_x"]**2 +
    project_df["gyro_y"]**2 +
    project_df["gyro_z"]**2
)

project_df[["acc_magnitude", "gyro_magnitude"]].head()
```

```python
# Create a simple numeric activity code.
# Machine learning models often need numbers instead of text labels.
activity_map = {
    activity: code
    for code, activity in enumerate(sorted(project_df["activity"].unique()))
}

project_df["activity_code"] = project_df["activity"].map(activity_map)

print(activity_map)
project_df[["activity", "activity_code"]].head()
```

## Summary
Better features can help models learn patterns more easily.

Encoding prepares labels for machine learning.
