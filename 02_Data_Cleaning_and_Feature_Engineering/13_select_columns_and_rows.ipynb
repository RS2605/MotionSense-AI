# 13 — Select Columns and Rows

## Goal
Select useful columns and rows with simple Pandas commands.


```python
from pathlib import Path

import numpy as np
import pandas as pd

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
PROCESSED_DIR = DATA_DIR / "processed"

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Why selection matters
Real datasets often contain more columns than we need.

Selection lets us focus on the useful parts of the data.

```python
df = pd.read_csv(RAW_PATH)

# Select one column.
# A single column in Pandas is called a Series.
activities = df["activity"]
activities.head()
```

```python
# Select multiple columns by passing a list of column names.
# This creates a smaller DataFrame.
sensor_preview = df[["subject", "activity", "acc_x", "acc_y", "acc_z"]]
sensor_preview.head()
```

```python
# iloc selects by integer position.
# This means: first 5 rows and first 4 columns.
df.iloc[:5, :4]
```

```python
# loc selects by labels.
# This means: rows 0 through 4 and these named columns.
df.loc[0:4, ["row_id", "subject", "activity", "device"]]
```

## Summary
Use column names when meaning matters.

Use `iloc` when position matters.
