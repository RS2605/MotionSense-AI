# 17 — Clean Activity Text and Check Data Types

## Goal
To standardize text values in the activity column
and To confirm that the numeric columns are ready for calculations.

```python
from pathlib import Path

import numpy as np
import pandas as pd

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
PROCESSED_DIR = DATA_DIR / "processed"

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Why this matters
Pandas treats even slightly different text values as different categories.

For example, `Walking`, `walking`, and ` WALKING ` are not the same unless we clean them.

```python
df = pd.read_csv(RAW_PATH)

# Create a copy so the raw data stays unchanged.
clean_df = df.copy()

# Clean the activity text.
# strip() removes extra spaces.
# title() gives a consistent readable format.
clean_df["activity"] = clean_df["activity"].str.strip().str.title()

print(clean_df["activity"].unique())
```

```python
# Check data types.
# Sensor columns should be numeric so we can calculate statistics and features.
clean_df.dtypes
```

```python
# Convert sensor columns to numeric values.
# errors="coerce" turns invalid values into NaN instead of crashing.
sensor_columns = ["acc_x", "acc_y", "acc_z", "gyro_x", "gyro_y", "gyro_z"]

for column in sensor_columns:
    clean_df[column] = pd.to_numeric(clean_df[column], errors="coerce")

clean_df[sensor_columns].dtypes
```

## Summary
Clean text prevents duplicate categories.

Correct data types make calculations reliable.
