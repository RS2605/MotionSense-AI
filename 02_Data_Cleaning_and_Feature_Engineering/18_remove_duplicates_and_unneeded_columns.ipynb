# 18 — Remove Duplicates and Unneeded Columns

## Goal
To remove duplicate rows and create a smaller, more focused working dataset.

```python
from pathlib import Path

import numpy as np
import pandas as pd

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
PROCESSED_DIR = DATA_DIR / "processed"

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Why cleanup matters
A smaller, cleaner dataset is easier to analyze.

In this notebook, we remove duplicate rows and keep only useful columns.

```python
df = pd.read_csv(RAW_PATH)

# Count duplicate rows.
duplicate_count = df.duplicated().sum()
print("Duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows.
# If there are no duplicates, the DataFrame stays the same.
clean_df = df.drop_duplicates().reset_index(drop=True)

print("Original shape:", df.shape)
print("After duplicate removal:", clean_df.shape)
```

```python
# Keep only the columns needed for the MotionSense-AI project.
useful_columns = [
    "row_id", "subject", "time_step", "activity",
    "acc_x", "acc_y", "acc_z",
    "gyro_x", "gyro_y", "gyro_z",
    "device"
]

project_df = clean_df[useful_columns].copy()
project_df.head()
```

## Summary
Cleanup keeps the dataset focused and enhances later model training and performance.
