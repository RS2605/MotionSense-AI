# 15 — Sort Data and Reset Index

## Goal
To sort rows by sensor values and clean the row index after sorting.

```python
from pathlib import Path

import numpy as np
import pandas as pd

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
PROCESSED_DIR = DATA_DIR / "processed"

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Why sorting matters
Sorting helps us quickly find high or low sensor readings.

After sorting or filtering, resetting the index makes row numbers clean again.

```python
df = pd.read_csv(RAW_PATH)

# Sort rows by acc_z from highest to lowest.
# acc_z is one accelerometer feature.
sorted_df = df.sort_values(by="acc_z", ascending=False)

sorted_df[["row_id", "activity", "acc_z", "device"]].head()
```

```python
# The old row index is still attached after sorting.
# reset_index(drop=True) creates clean row numbers starting at 0.
sorted_clean_index = sorted_df.reset_index(drop=True)

sorted_clean_index[["row_id", "activity", "acc_z", "device"]].head()
```

```python
# Sort by activity first, then by time_step inside each activity.
activity_time_df = df.sort_values(by=["activity", "time_step"]).reset_index(drop=True)
activity_time_df.head()
```

## Summary
Sorting organizes rows. Resetting the index makes the result easier to read.
