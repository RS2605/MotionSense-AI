# 16 — Find and Handle Missing Values

## Goal
To identify missing sensor values and fill simple numeric missing values.

```python
from pathlib import Path

import numpy as np
import pandas as pd

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
PROCESSED_DIR = DATA_DIR / "processed"

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Missing data
Real-world data is rarely perfect.

Missing values should be detected and handled before analysis or machine learning.

```python
df = pd.read_csv(RAW_PATH)

# Count missing values in every column.
missing_counts = df.isnull().sum()

# Show only columns that actually have missing values.
missing_counts[missing_counts > 0]
```

```python
# Display rows where at least one value is missing.
# any(axis=1) checks across each row.
rows_with_missing = df[df.isnull().any(axis=1)]
rows_with_missing.head()
```

```python
# Make a copy before cleaning.
# This protects the original DataFrame.
clean_df = df.copy()

# Fill missing numeric sensor values with the column mean.
sensor_columns = ["acc_x", "acc_y", "acc_z", "gyro_x", "gyro_y", "gyro_z"]

for column in sensor_columns:
    if column in clean_df.columns:
        clean_df[column] = clean_df[column].fillna(clean_df[column].mean())

# Verify missing values are gone from the sensor columns.
clean_df[sensor_columns].isnull().sum()
```

## Summary
First find missing values. Then use a simple, explainable cleaning strategy.
