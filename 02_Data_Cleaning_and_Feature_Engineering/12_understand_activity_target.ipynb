# 12 — Understand the Activity Target

## Goal
Explore the activity column, which is the label the model will eventually predict.


```python
from pathlib import Path

import numpy as np
import pandas as pd

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
PROCESSED_DIR = DATA_DIR / "processed"

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Target variable
The **activity** column is the answer label.

Later, the machine learning model will use sensor features to predict this activity.

```python
df = pd.read_csv(RAW_PATH)

# Show the first few activity labels.
# This confirms the target column exists and looks correct.
df["activity"].head()
```

```python
# value_counts() counts how many rows belong to each activity.
# This helps us check whether the dataset is balanced.
activity_counts = df["activity"].value_counts()
print(activity_counts)
```

```python
# nunique() counts the number of unique activity classes.
num_activities = df["activity"].nunique()
print("Number of activity classes:", num_activities)

# unique() shows the actual activity names.
print("Activity names:")
print(df["activity"].unique())
```

## Summary
A balanced target means each activity has a similar number of examples, which helps later model training.
