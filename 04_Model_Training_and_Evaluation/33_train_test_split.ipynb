# 33 — Train-Test Split

## Goal
To split the dataset into training data and testing data.

## Research Question
How can we test: how well the model works on data it has not seen before?

## Why split the data?

A model should be evaluated on examples it did not memorize during training.

```text
Full Dataset → Training Set + Testing Set
```

````python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data")
PROCESSED_DIR = DATA_DIR / "processed"
ML_READY_PATH = PROCESSED_DIR / "motionsense_ml_ready.csv"

from sklearn.model_selection import train_test_split

df = pd.read_csv(ML_READY_PATH)

feature_columns = [
    "acc_x", "acc_y", "acc_z",
    "gyro_x", "gyro_y", "gyro_z",
    "acc_magnitude", "gyro_magnitude"
]

X = df[feature_columns]
y = df["activity"]

# test_size=0.20 means 20% of the data is kept for testing.
# random_state=42 makes the split repeatable.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
````

````python
# Check that each of the activity classes is represented proportionally in both training and testing data.
print("Training activity counts:")
print(y_train.value_counts())

print("\nTesting activity counts:")
print(y_test.value_counts())
````

## Key idea

The model learns from `X_train` and `y_train`, then proves itself on `X_test` and `y_test`.
