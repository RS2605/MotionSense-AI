# 36 — Measure Accuracy

## Goal
To calculate how often the model predicts the correct activity.

## Research Question
How accurate is the trained MotionSense-AI model?

## Accuracy

Accuracy is the percentage of predictions the model gets correct.

```text
Accuracy = Correct Predictions / Total Predictions
```

````python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data")
PROCESSED_DIR = DATA_DIR / "processed"
ML_READY_PATH = PROCESSED_DIR / "motionsense_ml_ready.csv"

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

df = pd.read_csv(ML_READY_PATH)
feature_columns = [
    "acc_x", "acc_y", "acc_z",
    "gyro_x", "gyro_y", "gyro_z",
    "acc_magnitude", "gyro_magnitude"
]
X = df[feature_columns]
y = df["activity"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

model = RandomForestClassifier(n_estimators=50, random_state=42)
model.fit(X_train, y_train)
predictions = model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)
print(f"Accuracy: {accuracy:.2%}")
````

````python
# Simple baseline: most common activity in the training data (i.e. the mode value in the training data).
# A model should perform better than this simple rule.
most_common_activity = y_train.value_counts().idxmax()
baseline_predictions = [most_common_activity] * len(y_test)
baseline_accuracy = accuracy_score(y_test, baseline_predictions)

print("Most common activity:", most_common_activity)
print(f"Baseline accuracy: {baseline_accuracy:.2%}")
print(f"Model accuracy: {accuracy:.2%}")
````

## Key idea

A model is meaningful only if it performs better than a simple baseline.
