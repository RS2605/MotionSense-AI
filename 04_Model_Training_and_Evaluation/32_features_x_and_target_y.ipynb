# 32 — Features X and Target y

## Goal
To separate the input features from the output label.

## Research Question
Which columns should teach the model, and which column should the model predict?

## Machine learning vocabulary

- **Features (`X`)** are the inputs.
- **Target (`y`)** is the answer we want to predict.

For this project, `activity` is the target.

```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data")
PROCESSED_DIR = DATA_DIR / "processed"
ML_READY_PATH = PROCESSED_DIR / "motionsense_ml_ready.csv"

df = pd.read_csv(ML_READY_PATH)

# Choose a small, clear set of input features.
# Keeping the feature list focused results in better model training.
feature_columns = [
    "acc_x", "acc_y", "acc_z",
    "gyro_x", "gyro_y", "gyro_z",
    "acc_magnitude", "gyro_magnitude"
]

# X contains the inputs used for prediction.
X = df[feature_columns]

# y contains the correct activity labels.
y = df["activity"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## Key idea

Never put the target column inside `X`. If the model sees the answer during training, the evaluation will not be honest.
