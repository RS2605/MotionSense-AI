# 38 — Classification Report

## Goal
To review precision, recall, and F1-score for each activity.

## Research Question
Does the model perform equally well for every activity?

## Beyond accuracy

Accuracy is helpful, but it does not show performance for each activity. A classification report gives a more detailed insight.

```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data")
PROCESSED_DIR = DATA_DIR / "processed"
ML_READY_PATH = PROCESSED_DIR / "motionsense_ml_ready.csv"

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

df = pd.read_csv(ML_READY_PATH)
feature_columns = [
    "acc_x", "acc_y", "acc_z",
    "gyro_x", "gyro_y", "gyro_z",
    "acc_magnitude", "gyro_magnitude"
]
X = df[feature_columns]
y = df["activity"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

model = RandomForestClassifier(n_estimators=50, random_state=42)
model.fit(X_train, y_train)
predictions = model.predict(X_test)

# Print detailed metrics for each activity class.
print(classification_report(y_test, predictions))
```

## How to read the report

- **Precision:** When the model predicts an activity, how often is it correct?
- **Recall:** Of all true examples of an activity, how many did the model find?
- **F1-score:** A balanced score combining precision and recall.
- **Support:** Number of examples in the test set.
