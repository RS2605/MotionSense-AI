## Why this matters

Machine learning should not start from messy raw data. By now, the dataset is cleaned, feature engineered, and ready for training.

````python
from pathlib import Path
import pandas as pd

# DATA_DIR points to the shared data folder for this project.
DATA_DIR = Path("../data")
PROCESSED_DIR = DATA_DIR / "processed"
ML_READY_PATH = PROCESSED_DIR / "motionsense_ml_ready.csv"

# Load the ML-ready dataset.
df = pd.read_csv(ML_READY_PATH)

# Display the size of the dataset: rows and columns.
print("Dataset shape:", df.shape)

# Preview the first few rows.
df.head()
````

````python
# Display all column names so we know what inputs are available.
print(df.columns.tolist())
````

````python
# Check the activity labels.
# These are the classes the model will learn to predict.
print(df["activity"].unique())
print("Number of activities:", df["activity"].nunique())
````

## Key idea

The model will use sensor columns such as `acc_x`, `acc_y`, `gyro_x`, and engineered features such as `acc_magnitude` to predict the `activity` column.
