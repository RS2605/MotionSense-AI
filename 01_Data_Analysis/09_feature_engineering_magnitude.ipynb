# 09 — Feature Engineering: Magnitude

**Goal:** To create useful new sensor features from existing columns.

```python
from pathlib import Path
import numpy as np
import pandas as pd

DATA_PATH = Path('../data/raw/motionsense_raw.csv')

# Load the raw MotionSense-style sensor dataset.
df = pd.read_csv(DATA_PATH)
# Fill missing sensor values first so calculations work cleanly.
project_df = df.copy()
sensor_cols = ['acc_x', 'acc_y', 'acc_z', 'gyro_x', 'gyro_y', 'gyro_z']
for col in sensor_cols:
    project_df[col] = (project_df[col].fillna(project_df[col].mean()))

project_df.head()
```

## What is Magnitude?

The phone measures motion in three directions:

- X direction
- Y direction
- Z direction

Magnitude combines the three directions into one overall movement value.

```python
# Create acceleration magnitude.
# This combines acc_x, acc_y, and acc_z into one overall movement feature.
project_df['acc_magnitude'] = np.sqrt(
    project_df['acc_x']**2 +
    project_df['acc_y']**2 +
    project_df['acc_z']**2
)

project_df[['acc_x', 'acc_y', 'acc_z', 'acc_magnitude']].head()
```

```python
# Create gyroscope magnitude.
# This combines gyro_x, gyro_y, and gyro_z into one overall rotation feature.
project_df['gyro_magnitude'] = np.sqrt(
    project_df['gyro_x']**2 +
    project_df['gyro_y']**2 +
    project_df['gyro_z']**2
)

project_df[['gyro_x', 'gyro_y', 'gyro_z', 'gyro_magnitude']].head()
```

```python
# Compare average acceleration magnitude by activity.
# This gives a first look at whether activities have different movement patterns.
project_df.groupby('activity')['acc_magnitude'].mean().sort_values(ascending=False)
```

## Summary

Feature engineering creates new columns that may help a machine learning model.

Here we created:

- `acc_magnitude` = overall acceleration
- `gyro_magnitude` = overall rotation
