# 05 — Read and Inspect Data with Pandas

**Goal:** To use Pandas to load and inspect the shared raw dataset.

```python
from pathlib import Path
import numpy as np
import pandas as pd

# All notebooks use the same raw dataset.
DATA_PATH = Path('../data/raw/motionsense_raw.csv')

# Load the raw MotionSense-style sensor dataset.
df = pd.read_csv(DATA_PATH)

# head() displays the first five rows.
# This is usually the first command used after loading a dataset.
df.head()
```

```python
# shape tells us the dataset size.
print('Rows and columns:', df.shape)

# columns shows all column names.
print(df.columns.tolist())
```

```python
# info() summarizes column names, non-missing values, and data types.
# This helps us check whether the data looks reasonable.
df.info()
```

```python
# describe() summarizes numerical columns.
# It shows count, mean, standard deviation, min, and max.
df.describe()
```

## Summary

As a good practice, we use these four commands whenever we load a new dataset:

- `head()`
- `shape`
- `info()`
- `describe()`
