# 02 — NumPy Arrays with Sensor Values

**Goal:** To learn how NumPy stores sensor numbers in arrays.

```python
import numpy as np

# A simple list of accelerometer readings from one sensor direction.
python_list = [0.12, 0.18, 0.09, 0.21, 0.15]

# Convert the Python list into a NumPy array.
# NumPy arrays make numerical calculations easy and fast.
sensor_values = np.array(python_list)

print(sensor_values)
print(type(sensor_values))
```

```python
# Basic calculations on the full array.
# NumPy applies these functions to all values in the array.
print('Mean:', sensor_values.mean())
print('Minimum:', sensor_values.min())
print('Maximum:', sensor_values.max())
```

```python
# Array structure tells us how the data is organized.
print('Number of dimensions:', sensor_values.ndim)
print('Shape:', sensor_values.shape)
print('Total values:', sensor_values.size)
print('Data type:', sensor_values.dtype)
```

## Summary

A NumPy array is a clean way to store many numbers together.

For sensor data, arrays are useful because we often calculate averages, minimums, maximums, and other summaries.
