# 04 — Vectorized Calculations

**Goal:** To apply one calculation to an entire array without writing a loop.

```python
import numpy as np

acc_x = np.array([0.10, 0.20, 0.30, 0.40])
print(acc_x)
```

```python
# Add a constant to every value.
# This happens element by element.
adjusted = acc_x + 0.05
print(adjusted)
```

```python
# Multiply every value by 2.
doubled = acc_x * 2
print(doubled)
```

```python
# Square every value.
# Squaring is useful when calculating magnitude later.
squared = acc_x ** 2
print(squared)
```

## Summary

Vectorization means: **one operation applies to the whole array**.

This keeps code short and readable.
