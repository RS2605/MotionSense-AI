# 03 — Indexing and Slicing Sensor Windows

**Goal:** To select individual values and small windows from sensor data.

```python
import numpy as np

# Ten sensor readings collected over time.
readings = np.array([10, 12, 13, 15, 18, 17, 16, 14, 13, 11])
print(readings)
```

```python
# Indexing selects one value.
# Python starts counting at 0, so readings[0] is the first value.
print('First reading:', readings[0])
print('Third reading:', readings[2])
print('Last reading:', readings[-1])
```

```python
# Slicing selects multiple values.
# The end position is excluded.
print('First 5 readings:', readings[:5])
print('Readings from position 2 to 5:', readings[2:6])
print('Last 3 readings:', readings[-3:])
```

```python
# A small sensor window is a short sequence of readings.
# Windows are important because activities are patterns over time.
window = readings[:5]
print('Sensor window:', window)
print('Window average:', window.mean())
```

## Summary

- **Indexing** selects one value.
- **Slicing** selects a range of values.
- Sensor windows help us study movement over time.
