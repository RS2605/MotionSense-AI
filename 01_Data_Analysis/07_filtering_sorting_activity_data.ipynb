# 07 — Filtering and Sorting Activity Data

**Goal:** To keep only rows that match a condition.

```python
from pathlib import Path
import numpy as np
import pandas as pd

DATA_PATH = Path('../data/raw/motionsense_raw.csv')

df = pd.read_csv(DATA_PATH)
df.head()
```

```python
# Count how many rows belong to each activity.
# This helps us understand the target column.
df['activity'].value_counts()
```

```python
# Keep only Walking rows.
walking_df = df[df['activity'] == 'Walking']

print('Walking rows:', walking_df.shape[0])
walking_df.head()
```

```python
# Keep only rows where acc_z is greater than 1.2.
# This is a simple numerical filter.
high_acc_z = df[df['acc_z'] > 1.2]

print('Rows with high acc_z:', high_acc_z.shape[0])
high_acc_z.head()
```

```python
# Sort rows by acc_z from highest to lowest.
# This helps us quickly see the largest values.
sorted_df = df.sort_values(by='acc_z', ascending=False)
sorted_df.head()
```

## Summary

Filtering answers questions like:

- Show only Walking rows.
- Show rows with high acceleration.
- Sort values to find largest or smallest readings.
