# 25 — Line Chart of Sensor Signals

## Goal
To plot sensor readings over time for a small window of rows.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
CLEAN_PATH = DATA_DIR / "processed" / "motionsense_clean.csv"
ML_READY_PATH = DATA_DIR / "processed" / "motionsense_ml_ready.csv"
FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(exist_ok=True)

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Why line charts matter
Sensor data changes over time. A line chart helps us see those changes.

We will plot only a small window so the chart stays readable.

```python
df = pd.read_csv(CLEAN_PATH)

# Select the first 100 rows from one activity.
# This keeps the chart simple and easy to interpret.
walking = df[df["activity"] == "Walking"].head(100)

print("Rows selected:", walking.shape[0])
walking[["time_step", "activity", "acc_x", "acc_y", "acc_z"]].head()
```

```python
# Plot three accelerometer axes on one chart.
plt.figure(figsize=(10, 4))
plt.plot(walking["time_step"], walking["acc_x"], label="acc_x")
plt.plot(walking["time_step"], walking["acc_y"], label="acc_y")
plt.plot(walking["time_step"], walking["acc_z"], label="acc_z")
plt.title("Accelerometer Signals for Walking")
plt.xlabel("Time Step")
plt.ylabel("Acceleration")
plt.legend()
plt.tight_layout()
plt.show()
```

## Summary
Line charts are useful for viewing sensor signals over time.
