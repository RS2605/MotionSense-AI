# 24 — Histogram of Motion Distribution

## Goal
To use a histogram to see how motion values are distributed.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
CLEAN_PATH = DATA_DIR / "processed" / "motionsense_clean.csv"
ML_READY_PATH = DATA_DIR / "processed" / "motionsense_ml_ready.csv"
FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(exist_ok=True)

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Histogram idea
A histogram shows where most numeric values are located.

For MotionSense-AI, we can use histograms to inspect overall acceleration and rotation.

```python
df = pd.read_csv(CLEAN_PATH)

# Create a histogram of accelerometer magnitude.
plt.figure(figsize=(8, 4))
plt.hist(df["acc_magnitude"], bins=20)
plt.title("Distribution of Accelerometer Magnitude")
plt.xlabel("acc_magnitude")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# Create a histogram of gyroscope magnitude.
plt.figure(figsize=(8, 4))
plt.hist(df["gyro_magnitude"], bins=20)
plt.title("Distribution of Gyroscope Magnitude")
plt.xlabel("gyro_magnitude")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

## Summary
Histograms help us see the shape of numeric data and identify unusual values.
