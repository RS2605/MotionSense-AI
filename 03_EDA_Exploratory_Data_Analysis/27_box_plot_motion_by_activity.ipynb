# 27 — Box Plot of Motion by Activity

## Goal
To use a box plot to compare the spread of motion values across activities.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
CLEAN_PATH = DATA_DIR / "processed" / "motionsense_clean.csv"
ML_READY_PATH = DATA_DIR / "processed" / "motionsense_ml_ready.csv"
FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(exist_ok=True)

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Box plot idea
A box plot shows how values are spread out for each group.

It is useful for comparing activities because each activity may have a different movement pattern.

```python
df = pd.read_csv(CLEAN_PATH)

# Create a box plot comparing accelerometer magnitude by activity.
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="activity", y="acc_magnitude")
plt.title("Accelerometer Magnitude by Activity")
plt.xlabel("Activity")
plt.ylabel("acc_magnitude")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
# Create a box plot comparing gyroscope magnitude by activity.
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="activity", y="gyro_magnitude")
plt.title("Gyroscope Magnitude by Activity")
plt.xlabel("Activity")
plt.ylabel("gyro_magnitude")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

## Summary
Box plots help us compare activity groups and spot unusual sensor values.
