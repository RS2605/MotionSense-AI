# 23 — Activity Counts and Class Balance

## Goal
To count number of activity labels in each unique activity category and check whether the dataset is balanced.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
CLEAN_PATH = DATA_DIR / "processed" / "motionsense_clean.csv"
ML_READY_PATH = DATA_DIR / "processed" / "motionsense_ml_ready.csv"
FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(exist_ok=True)

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Why class balance matters
The **activity** column is the target we wish to predict.

If one activity has many more examples than others, the model may become biased towards that activity.

```python
df = pd.read_csv(CLEAN_PATH)

# Count how many observations belong to each activity.
activity_counts = df["activity"].value_counts()
activity_counts
```

```python
# Convert counts into percentages.
activity_percentages = df["activity"].value_counts(normalize=True) * 100
activity_percentages.round(2)
```

```python
# Create a simple bar chart using Pandas.
activity_counts.plot(kind="bar", figsize=(8, 4))
plt.title("Activity Counts")
plt.xlabel("Activity")
plt.ylabel("Number of Observations")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

## Summary
Counting labels helps us understand whether the dataset gives each activity enough examples.
