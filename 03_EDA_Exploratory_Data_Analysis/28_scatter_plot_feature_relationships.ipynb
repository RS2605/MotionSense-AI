# 28 — Scatter Plot Feature Relationships

## Goal
To use scatter plots to compare two numeric features and color points by activity.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
CLEAN_PATH = DATA_DIR / "processed" / "motionsense_clean.csv"
ML_READY_PATH = DATA_DIR / "processed" / "motionsense_ml_ready.csv"
FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(exist_ok=True)

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Scatter plot idea
A scatter plot compares two numeric features.

If activities form different clusters, that suggests the features may help a machine learning model.

```python
df = pd.read_csv(CLEAN_PATH)

# Use a smaller sample so the plot is easier to read and analyze.
plot_df = df.sample(n=300, random_state=42)

plot_df[["activity", "acc_magnitude", "gyro_magnitude"]].head()
```

```python
# Compare accelerometer magnitude and gyroscope magnitude.
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=plot_df,
    x="acc_magnitude",
    y="gyro_magnitude",
    hue="activity"
)
plt.title("acc_magnitude vs gyro_magnitude")
plt.xlabel("Accelerometer Magnitude")
plt.ylabel("Gyroscope Magnitude")
plt.tight_layout()
plt.show()
```

## Summary
Scatter plots reveal relationships between features and can show whether activities separate visually.
