# 22 — Descriptive Statistics

## Goal
To use basic statistics to summarize sensor features.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_DIR = Path("../data")
RAW_PATH = DATA_DIR / "raw" / "motionsense_raw.csv"
CLEAN_PATH = DATA_DIR / "processed" / "motionsense_clean.csv"
ML_READY_PATH = DATA_DIR / "processed" / "motionsense_ml_ready.csv"
FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(exist_ok=True)

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## Statistics answer simple questions
Statistics help us answer questions such as:

- What is the average sensor reading?
- What is the smallest value?
- What is the largest value?
- How spread out are the values?

```python
df = pd.read_csv(CLEAN_PATH)

# Select a small set of important numeric features.
features = ["acc_x", "acc_y", "acc_z", "acc_magnitude", "gyro_magnitude"]

# describe() gives count, mean, standard deviation, min, quartiles, and max.
df[features].describe().round(3)
```

```python
# Calculate individual statistics for one feature.
# acc_magnitude represents overall acceleration.
print("Mean:", df["acc_magnitude"].mean())
print("Min:", df["acc_magnitude"].min())
print("Max:", df["acc_magnitude"].max())
print("Std:", df["acc_magnitude"].std())
```

## Summary
`describe()` is a quick way to understand numeric features before creating charts.
